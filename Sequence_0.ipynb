## 4.8 Notion de complexité
Nous considérons désormais un autre algorithme permettant le calcul de la puissance d'un nombre. En voici une implémentation :

```python
def puissance2(x,n):
    R=1
    X=x
    N=n
    while (N>0):
        if ((N%2)!=0):
            R=R*X
        N=N//2
        X=X*X
    return R
```

### Calcul du nombre d'opérations
Nous souhaitons comparer le nombre d'opérations effectuées par les programmes puissance1 et puissance2 en fonction de la valeur de n. 
<br>La première étape consiste à créer la fonction permettant de compter le nombre d'opérations de chacun des programmes.

```python
# using the power2 function to create a counter for the number of instructions performed by that function
# we will use a counter j
def puissance2complex(x,n):
    R=1
    X=x
    N=n
    # we initialize counter j at 0 
    j=0     
    while (N>0):
        if ((N%2)!=0):
            R=R*X
        N=N//2
        X=X*X
        #  a maximum of 4 operations are performed for each call in the loop, so we increment j accordingly 
        j = j+4
    # we return the value of the counter
    return j
```
