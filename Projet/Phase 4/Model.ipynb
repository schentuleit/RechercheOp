```python
import sys, time, itertools, numpy as np
sys.path.insert(0, r'C:\Users\schen\Documents\CESI\RechercheOP\RechercheOp\Projet\Phase 4')

for mod in ['env', 'model', 'train', 'preprocess']:
    if mod in sys.modules:
        del sys.modules[mod]

from preprocess import generate_instance, preprocess
from train      import run_episode, load_model
import torch

# ── Charger le modèle n=20 (généralise bien sur n=10) ────────────────
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model, _, _ = load_model(
    r'C:\Users\schen\Documents\CESI\RechercheOP\RechercheOp\Projet\Phase 4\checkpoints_n20\model_best.pt'
)
model.eval()

# ── Brute force VRPTW ────────────────────────────────────────────────
def eval_route(route, instance):
    """Évalue le coût d'une tournée pour UN véhicule."""
    if not route:
        return 0.0
    coords   = instance['coords']
    durees   = instance['durees']
    tw       = instance['time_windows']
    service  = instance['service_times']
    capacity = instance['capacity']
    demands  = instance['demands']

    # Vérif capacité
    if sum(demands[i] for i in route) > capacity + 1e-6:
        return float('inf')

    t    = 0.0
    pos  = 0   # dépôt
    dist = 0.0

    for ville in route:
        t_arrivee = t + durees[pos, ville]
        t_arrivee = max(t_arrivee, tw[ville, 0])
        if t_arrivee > tw[ville, 1] + 2.0:
            return float('inf')   # fenêtre violée
        t    = t_arrivee + service[ville]
        dist += np.linalg.norm(coords[pos] - coords[ville])
        pos   = ville

    # Retour dépôt
    if t + durees[pos, 0] > instance['horizon'] + 1e-6:
        return float('inf')
    dist += np.linalg.norm(coords[pos] - coords[0])
    return dist


def brute_force(instance):
    """
    Brute force sur n=10 — teste toutes les répartitions possibles
    des clients sur 3 véhicules puis toutes les permutations par véhicule.
    Limité à n<=10 pour rester raisonnable.
    """
    n       = instance['n']
    clients = list(range(1, n + 1))
    K       = instance['n_vehicles']
    best    = float('inf')
    best_routes = None

    # Générer toutes les partitions de clients en K groupes
    # Via assignation : chaque client → véhicule 0, 1 ou 2
    for assignation in itertools.product(range(K), repeat=n):
        groupes = [[] for _ in range(K)]
        for i, k in enumerate(assignation):
            groupes[k].append(clients[i])

        # Pour chaque groupe, tester toutes les permutations
        cout_total = 0.0
        routes     = []
        valide     = True

        for groupe in groupes:
            if not groupe:
                routes.append([])
                continue
            meilleur_cout  = float('inf')
            meilleure_perm = None
            for perm in itertools.permutations(groupe):
                c = eval_route(list(perm), instance)
                if c < meilleur_cout:
                    meilleur_cout  = c
                    meilleure_perm = list(perm)
            if meilleur_cout == float('inf'):
                valide = False
                break
            cout_total += meilleur_cout
            routes.append(meilleure_perm)

        if valide and cout_total < best:
            best        = cout_total
            best_routes = routes

    return best, best_routes


# ── Comparaison sur 5 instances n=10 ────────────────────────────────
print(f"{'='*65}")
print(f"  Comparaison Brute Force vs Modèle DL — n=10, 3 véhicules")
print(f"{'='*65}")
print(f"  {'Seed':>6}  {'Brute Force':>12}  {'Modèle DL':>10}  {'Gap':>8}  {'tps BF':>8}  {'tps DL':>8}")
print(f"  {'-'*63}")

for seed in range(2000, 2005):
    inst = generate_instance(10, n_vehicles=3, seed=seed)
    feat = preprocess(inst)

    # Brute force
    t0 = time.time()
    bf_dist, bf_routes = brute_force(inst)
    t_bf = time.time() - t0

    # Modèle DL
    t0 = time.time()
    with torch.no_grad():
        dl_dist, _, _ = run_episode(model, inst, feat, device, greedy=True)
    t_dl = time.time() - t0

    gap = (dl_dist - bf_dist) / bf_dist * 100 if bf_dist > 0 else 0

    print(f"  {seed:>6}  {bf_dist:>12.1f}  {dl_dist:>10.1f}  {gap:>+7.1f}%  {t_bf:>7.1f}s  {t_dl*1000:>6.1f}ms")

print(f"  {'='*63}")
print(f"\n  Gap positif = modèle moins bon que BF")
print(f"  Gap négatif = modèle meilleur que BF (improbable mais possible si BF manque des solutions valides)")
```
